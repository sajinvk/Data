# 1-Lists     

```python
l1 = ['RAG','is','awesome']
print(f"Orignial list: {l1}")
#adding an item 
l1.append('!')
print(f"List after adding !: {l1}")

#Removing an item 
l1.remove('awesome')
print(f"List after removing 'awesome': {l1}")
```

## list with comprehension 

```python
#Comprehension to create a list of squares 

squares = [x**2 for x in range(10)]
print(f"Squares of numbers 0 to 9: {squares}")

# Same thing without comprehension instead using for loop 

squares_for_loop =  []

for x in range(11,19):
    squares_for_loop.append(x**2)

print(f"Squares of numbers 0 to 9: {squares_for_loop}")
```

```python
# Conditional List comprehension 

even_squares = [x**2 for  x in range(10) if x%2 == 0]
print(f"squares of even numbers for 0 to 9: {even_squares} with list comprehension")

even_squares_for_loop =  []
for x in range(10):
    if x%2 == 0:
        even_squares_for_loop.append(x**2)
print(f"Squares of even numbers done using for loop and without comprehension : {even_squares_for_loop}")
```

# 2 - Dictionaries  

```python
person = {
    'name' : 'Alice',
    'age' : 24,
    'city' : 'New York'
}
print(f"person dict: {person}")
```

```python
for key, value in person.items():
    print(f"key: {key}\tValue: {value}")
```

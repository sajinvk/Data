```python
import os
import truststore

os.environ["AWS_PROFILE"] = "NP-DP-SysAccounts-AllPlatforms"
os.environ["AWS_REGION"] = "ap-southeast-2"
os.environ["AWS_DEFAULT_REGION"] = "ap-southeast-2"

truststore.inject_into_ssl()

from anthropic import AnthropicBedrock

client = AnthropicBedrock(
    aws_region="ap-southeast-2"
)

model = "au.anthropic.claude-haiku-4-5-20251001-v1:0"
```

```python
response = client.messages.create(
    model=model,
    max_tokens=100,
    messages=[
        {
            "role": "user",
            "content": "Things to remember about messages.create for anthropic client"
        }
    ]
)

print(response.content[0].text)
print("Stop reason:", response.stop_reason)
print("Usage:", response.usage)
```

Multi Turn Conversation 

```python
def add_user_message(messages,text):
    user_message = {'role':'user','content':text}
    messages.append(user_message)

def add_assistant_message(messages,text):
    assistant_message = {'role':'assistant','content':text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model=model,
        max_tokens=100,
        messages=messages
    )
    return message.content[0].text 
```

```python
# Empty Message 
messages = []

#add user message 
add_user_message(messages, "Is there any other role user , assistant while making API calls to claude")

#Get Claude's response
answer = chat(messages)
print(f"1st Answer: {answer}")

# Add Claude's response to the conversation history
add_assistant_message(messages, answer)

# Add a follow-up question
add_user_message(messages, "explain in one line")

# Get the follow-up response with full context
final_answer = chat(messages)
print(f"Final Answer: {final_answer}")
```
